## Stage 2 – Data Cleaning and Pre-processing

**Handling Missing Values**

```python
#Fill numerical columns using the median
num_cols = ['annual_income', 'credit_score', 'monthly_spend']
for col in num_cols:
  df[col] = df[col].fillna(df[col].median())

# Fill categorical columns using the mode
cat_cols = ['education', 'occupation', 'city']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

| Skewness             | Meaning                 | Suggested action                             |
| -------------------- | ----------------------- | -------------------------------------------- |
| Between -0.5 and 0.5 | Approximately symmetric | No transformation usually needed             |
| -1 to -0.5           | Moderately left-skewed  | Review distribution; transformation may help |
| 0.5 to 1             | Moderately right-skewed | Consider transformation if modeling          |
| Less than -1         | Strongly left-skewed    | Investigate low-end extreme values           |
| Greater than 1       | Strongly right-skewed   | Check high-end outliers; consider log1p()    |

| Kurtosis       | Meaning                                              |
| -------------- | ---------------------------------------------------- |
| Around 0       | Tail shape similar to a normal distribution          |
| Greater than 0 | Heavy tails; extreme values/outliers are more likely |
| Greater than 3 | Very heavy tails; investigate outliers carefully     |
| Less than 0    | Flatter distribution with lighter tails              |

**Find outliers with IQR Method**

- IQR=Q3-Q1
- Lower Bound = Q1-1.5*IQR
- Upper Bound = Q3+1.5*IQR

```python
#IQR-based outlier counts
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {count} outliers")
```

```python
q1 = df['monthly_spend'].quantile(0.25)
q3 = df['monthly_spend'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

monthly_spend_outliers = df[
    (df['monthly_spend'] < lower_bound) |
    (df['monthly_spend'] > upper_bound)
]

monthly_spend_outliers
```

```python
q1 = df['avg_transaction_amount'].quantile(0.25)
q3 = df['avg_transaction_amount'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

avg_transaction_amount_outliers = df[
    (df['avg_transaction_amount'] < lower_bound) |
    (df['avg_transaction_amount'] > upper_bound)
]
avg_transaction_amount_outliers
```
